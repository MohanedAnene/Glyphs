```python
import math
import random
import numpy as np

# Function to generate a random color
def random_color():
    return f"#{random.randint(0, 0xFFFFFF):06x}"

# Function to generate a scaled star with a random outline color
def simple_scaled_star(x: float, canvas: mg.Canvas) -> None:
    # Translate the star vertically based on the input x
    canvas.tr.translate(0, mg.lerp(x, 0, 0.05))

    # Calculate the radius of the star based on the input x
    radius = mg.lerp(x, 0.01, canvas.ysize / 2)

    # Generate the vertices of the star
    vertices = []
    for segment in range(5):
        vertices.append(mg.orbit(canvas.center, segment * 2 * math.pi / 5, radius))
        vertices.append(mg.orbit(canvas.center, (segment + 0.5) * 2 * math.pi / 5,
                         math.cos(2 * math.pi / 5) / math.cos(math.pi / 5) * radius))

    # Draw the star with a random outline color
    canvas.lines(vertices, closed=True, width='10p', linecap='round', color=color)

# Generate and export stars 30 times
for i in range(5):

  color1 = random_color()
  color = random_color()

  while color1 == color:
    color = random_color()

    
  xvalues = np.random.uniform(0.0, 100.0, 120)

    # Export the star with a random outline color
  mg.export(simple_scaled_star, xvalues=xvalues,
            name=f"Random Colored Star {i+1}", short_name=f'star_{i+1}',
            path=os.path.join(outpath, f"Random Colored Star {i+1}.mglyph"),
            author="Mohaned Anene BARKALLAH", email="Mohanedanene.barkallah@gmail.com", version="1.0.0")
  color1 = color

  print(f"Exported star {i+1} with random outline color")
```
